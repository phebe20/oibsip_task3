### Standardising "Rating Count"

The "Rating Count" column contains several different representations of review counts. Some values use commas, some use the "K" abbreviation, and some contain text such as "Not enough ratings".

Before converting the column to a numeric type, these formats need to be standardised so that the resulting values represent the actual number of ratings as consistently as possible. Text placeholders that do not provide a usable count will be treated as missing.


```python
amazon_data_copy.loc[
    amazon_data_copy["Rating Count"].astype("string").str.contains("K", na=False),
    "Rating Count"
].head()
```

```python
# Start by treating the rating counts as text so the different formats can be cleaned
rating_count_check = (
    amazon_data_copy["Rating Count"]
    .astype("string")
    .str.strip()
    .str.replace(",", "", regex=False)
)
```

```python
def convert_rating_count(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if value.lower().endswith("k"):
        return float(value[:-1]) * 1000

    if value.replace(".", "", 1).isdigit():
        return float(value)

    return np.nan
```

```python
rating_count_check = rating_count_check.apply(convert_rating_count)
```

```python
rating_count_check.head(10)
```

```python
rating_count_check.loc[[77, 94, 120, 155, 229]]
```

```python
rating_count_check.isna().sum()
```

```python
# Replace the mixed text formats with the cleaned numeric rating counts
amazon_data_copy["Rating Count"] = rating_count_check
```

### Recovering Missing Discount Percentages

The missing-value check showed that some products have an actual price and discounted price but no recorded discount percentage. Since the two prices are available, I can calculate the missing discount instead of using a general replacement value.

```python
# Calculate missing discounts from the actual and discounted prices
missing_discount_mask = amazon_data_copy["discount_percentage"].isna()

calculated_discount = (
    (amazon_data_copy["actual_price"] - amazon_data_copy["Discounted_Price (INR)"])
    / amazon_data_copy["actual_price"]
    * 100
).round(2)

discounts_filled = missing_discount_mask.sum()

amazon_data_copy.loc[missing_discount_mask, "discount_percentage"] = (
    calculated_discount.loc[missing_discount_mask]
)

discounts_filled
```

### Recovering Missing Discounted Prices

For records where the discounted price is missing but the actual price and discount are available, I can calculate the selling price using those two fields. This is preferable to filling the gaps with the median because the replacement can be linked directly to the product's recorded pricing information.


```python
# Find missing discounted prices that can be calculated from the other two fields
missing_discounted_mask = amazon_data_copy["Discounted_Price (INR)"].isna()

can_calculate_discounted = (
    missing_discounted_mask
    & amazon_data_copy["actual_price"].notna()
    & amazon_data_copy["discount_percentage"].notna()
)

calculated_discounted_price = (
    amazon_data_copy["actual_price"]
    * (1 - amazon_data_copy["discount_percentage"] / 100)
).round(2)

discounted_prices_filled = can_calculate_discounted.sum()

amazon_data_copy.loc[can_calculate_discounted, "Discounted_Price (INR)"] = (
    calculated_discounted_price.loc[can_calculate_discounted]
)

discounted_prices_filled
```

```python
# Find missing actual prices that can be recovered from the discounted price and discount
missing_actual_mask = amazon_data_copy["actual_price"].isna()

can_calculate_actual = (
    missing_actual_mask
    & amazon_data_copy["Discounted_Price (INR)"].notna()
    & amazon_data_copy["discount_percentage"].notna()
    & amazon_data_copy["discount_percentage"].lt(100).fillna(False)
)

calculated_actual_price = (
    amazon_data_copy["Discounted_Price (INR)"]
    / (1 - amazon_data_copy["discount_percentage"] / 100)
).round(2)

actual_prices_filled = can_calculate_actual.sum()

amazon_data_copy.loc[can_calculate_actual, "actual_price"] = (
    calculated_actual_price.loc[can_calculate_actual]
)

actual_prices_filled
```

## Outlier Detection

After cleaning the numerical fields and handling the missing values, I checked for potential outliers using the Interquartile Range (IQR) method.

I applied the check to the price, discount, rating, and rating-count columns. The purpose is to identify values that are unusually far from the middle of each distribution. A value flagged by this method is not automatically a mistake, so I will review the results before deciding whether to retain, cap, or remove any observations.


```python
# Columns to check for potential outliers
numeric_columns = [
    "Discounted_Price (INR)",
    "actual_price",
    "discount_percentage",
    "rating",
    "Rating Count"]

def get_iqr_summary(series):
    values = series.dropna()

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outlier_count = (
        (values < lower_limit) | (values > upper_limit)
    ).sum()

    return pd.Series({
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": outlier_count})


outlier_report = pd.DataFrame({
    column: get_iqr_summary(amazon_data_copy[column])
    for column in numeric_columns
}).T

outlier_report.round(2)
```

### Reviewing the Detected Outliers

The IQR check identified potential outliers in the price, rating, and rating-count columns. However, being outside the calculated limits does not automatically mean a value is incorrect.

I will review a few of the flagged records to see whether the values appear reasonable for the products involved. I will only change an outlier if there is evidence of a data-quality problem; otherwise, I will retain it rather than remove or cap a potentially valid observation.
